# Experiment 01: Basic Image Processing using OpenCV
## Aim: To read, display, and perform basic image processing operations on an image using OpenCV.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("OpenCV Version:", cv2.__version__)
print("NumPy Version :", np.__version__)
```

```python
# Read Image
img = cv2.imread("dog.jpg")

if img is None:
    print("Image not found!")
else:
    print("Image loaded successfully.")
    print("Shape :", img.shape)
    print("Data Type :", img.dtype)
```

```python
# Display Image (OpenCV)
cv2.imshow("Original Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,6))
plt.imshow(rgb)
plt.title("RGB Image")
plt.axis("off")
plt.show()
```

```python
# Details of Image
height, width, channels = img.shape

print("Height :", height)
print("Width  :", width)
print("Channels :", channels)
```

```python
# Access Pixel
print("Pixel Value at (100,100):")
print(img[100,100])
```

```python
# Modify Pixel

modified = img.copy()

# Change pixel at row 100, column 100 to green
modified[100, 100] = [0, 255, 0]

# Convert BGR to RGB for displaying with Matplotlib
modified_rgb = cv2.cvtColor(modified, cv2.COLOR_BGR2RGB)

# Display modified image
plt.figure(figsize=(8, 6))
plt.imshow(modified_rgb)
plt.title("Modified Pixel")
plt.axis("off")
plt.show()
```

```python
# Region of Interest (ROI / Crop)

roi = img[0:300, 100:300]
plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python

# Convert to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.figure(figsize=(6, 4))
plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Resize
resize = cv2.resize(img, (400, 400))
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(resize, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
# Flip
horizontal = cv2.flip(img, 1)
vertical = cv2.flip(img, 0)
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(horizontal, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(vertical, cv2.COLOR_BGR2RGB))
plt.title("Vertical Flip")
plt.axis("off")

plt.show()
```

```python
# Rotate
rotate = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(rotate, cv2.COLOR_BGR2RGB))
plt.title("Rotated 90° Clockwise")
plt.axis("off")
plt.show()
```

```python
# Draw Rectangle
rectangle = img.copy()

cv2.rectangle(rectangle,
              (100, 100),
              (400, 300),
              (0, 255, 0),
              3)
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(rectangle, cv2.COLOR_BGR2RGB))
plt.title("Rectangle")
plt.axis("off")
plt.show()
```

```python

# Add Text
text = img.copy()

cv2.putText(text,
            "23BIT066",
            (0, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (255, 255, 255),
            2)
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(text, cv2.COLOR_BGR2RGB))
plt.title("Text")
plt.axis("off")
plt.show()
```
